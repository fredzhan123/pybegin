```python
# 导入名为 Numpy 的第三方库，并将其重命名为 np
import numpy as np

# 调用 NumPy 库中的 arange() 函数,创建一个名为 x_array 的数组，包含从 0 到 9 共 10 个整数
x_array = np.arange(10)

# 打印数组
print(x_array)
```

```python
def my_function(x,y):
    """
    这个自定义函数计算两个数值 x 和 y 的和
    函数的输入为:
        x: 第一个数值
        y: 第二个数值
    函数的输出为:
        x + y: 两个数值的和
    """
    
    return x + y

print(my_function(1.5,2))
```
