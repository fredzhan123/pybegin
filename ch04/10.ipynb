# 列表索引和切片

```python
# 创建一个混合列表
my_list = [1,1.0,'1',True,[1,1.0,'1'],{1},{1:1.0}]
print(f'列表长度:{len(my_list)}')
```

```python
for index,item in enumerate(my_list):
    type_i = type(item)
    print(f'元素: {item},索引: {index},类型: {type_i}')
```

```python
# 列表索引
print(my_list[0])
print(my_list[1])
```

```python
print(my_list[-1])
print(my_list[-2])
```

```python
# 列表切片
# 取出前 3 个元素,索引为 0、1、2
print(my_list[:3])
```

```python
# 取出3个元素，不含0、不含 4
print(my_list[1:4])
```

```python
# 指定步长伟2，取出 0 、2、4、6 元素
print(my_list[::2])
```

```python
# 指定步长 -1， 倒序
print(my_list[::-1])
```

```python
# 提取列表中某个元素
print(my_list[4][1])
```
