# 将数据插入字符串

```python
name = 'James'
height = 181.18
```

```python
str_1 = name + ' has a height of ' + str(height) + ' cm.'
print(str_1)
```

```python
str_2 = '%s has a height of %.3f cm.' % (name,height)
print(str_2)
 
```

```python
str_3 = '{} has a height of {} cm.'.format(name,height)
print(str_3)
```

```python
str_4 = f'{name} has a height of {height} cm.'
print(str_4)
```
