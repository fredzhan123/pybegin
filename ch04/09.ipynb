# 常用 f-string 示例

```python
import datetime

now = datetime.datetime.now()
print("当前日期和时间是：", now)
print(f'现在的时间是: {now:%Y-%m-%d %H:%M:%S}')
print(f'现在的时间是: {now:%Y/%m/%d %H:%M:%S}')
```

```python
pi = 3.141592653589793238462643
f'{pi:.2f}'
```

```python
pi = 3.141592653589793238462643
f'{pi * 1000:.2e}'
```

```python
a = 18
# 转二进制
print(f"{a:b}")
```

```python
a = 68
# 转 十六进制
print(f"{a:x}")
```

```python
a = 88
# 转 八进制
print(f"{a:o}")
```
