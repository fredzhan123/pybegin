# 字符串的定义和操作

```python
str1 = "I am learning python 101"
print(str1)
```

```python
str2 = "Python is fun. Maching learning is fun too."
print(str2)
```

```python
# 使用加号 + 将多个字符串连接起来
str3 = 'Hey, ' + 'James!'
print(str3)
```

```python
# 使用乘号*将一个字符串复制多次
str4 = 'Python is Fun! '
str5 = str4 * 3
print(str5)
```

```python
# 字符串中的数字仅仅是字符
str6 = '123'
str7 = str6 * 3
print(str7)
```

```python
str8 = '456'
str9 = str6 + str8
print(str9)
print(type(str9))
```
